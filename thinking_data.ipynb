```python
import json
```

```python
# Load the data
def load_data(filename):
    with open (filename, "r") as f: 
        data = json.load(f)
        return data
```

```python
data = load_data("store_data.json")
print(data)
print(type(data))
```

```python
def clean_data(data):
    text_to_num = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5}
    cleaned_data = []
    unique_users = set()

    for user in data:
        # 1. Clean Rating
        raw_rating = str(user["rating"]).strip().lower()
        if raw_rating in text_to_num:
            raw_rating = text_to_num[raw_rating]
        user["rating"] = raw_rating

        # 2. Handle Missing Age
        raw_age = user.get("age")
        if raw_age == None:
            user["age"] = None

        # 3. Handle Duplication (Yeh if-block se bahar, for-loop ke level par hona chahiye)
        clean_name = user["name"].strip()
        if clean_name in unique_users:
            continue

        unique_users.add(clean_name)
        cleaned_data.append(user)

    # Return hamesha for-loop poora khatam hone ke baad bahar aata hai
    return cleaned_data
```

```python
clean_data(data)
```
