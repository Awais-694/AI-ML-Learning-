```python
import numpy as np 
```

```python
nums = [1,2,3,4,5]
sub_list = nums[1:3]
print(sub_list)
```

```python
sub_list[0] = 200
print(sub_list)
print(nums)
```

```python
arr = np.array([1,2,3,4,5])
sub_arr = arr[1:3]
print(sub_arr)
```

```python
sub_arr[0] = 200
print(sub_arr)
print(arr)
```
